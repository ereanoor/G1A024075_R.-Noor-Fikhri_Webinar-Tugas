```python
# ==============================================================================
# INFORMATICS PRACTITIONER LECTURE SERIES 2026 - UNIB
# IMPLEMENTASI MATERI 3: Geographic Information System (GIS) & Spatial Analytics
# Pemateri: Widi Primadepa, S.Kom. (CityCare Property, Christchurch, New Zealand)
# ==============================================================================

# 1. INSTALL LIBRARY GEOSPASIAL (Gunakan di Google Colab jika belum terinstall)
!pip install geopandas folium shapely -q

import geopandas as gpd
import folium
from shapely.geometry import Point, Polygon
import pandas as pd

# 2. ALUR DAN SPESIFIKASI PENGOLAHAN DATA GEOSPASIAL (GIS PIPELINE)
print("=" * 65)
print(" 🌐 ALUR PENGOLAHAN DATA SPASIAL & WEB MAPPING (GIS)")
print("=" * 65)
print(" 1. Ingest Data Vektor (Point, Line, Polygon)")
print(" 2. Proyeksi Sistem Koordinat (CRS - Coordinate Reference System)")
print(" 3. Spatial Query & Filter Atribut (Data Analytics)")
print(" 4. Rendering Interactive Web Map (Folium Framework)")
print("=" * 65 + "\n")

# 3. GENERATE SYNTHETIC SPATIAL DATASET (LOKASI SEKITAR UNIB BENGKULU)
# Koordinat Wilayah Bengkulu (Latitude & Longitude)
data_lokasi = {
    'Nama_Fasilitas': [
        'Rektorat Universitas Bengkulu',
        'Gedung Informatika UNIB',
        'Perpustakaan Utama UNIB',
        'Stadion UNIB',
        'Taman Budaya Bengkulu'
    ],
    'Kategori': ['Pendidikan', 'Akademik', 'Fasilitas', 'Olahraga', 'Publik'],
    'Kapasitas_Orang': [200, 500, 1000, 5000, 1500],
    'Latitude': [-3.7571, -3.7582, -3.7565, -3.7590, -3.7650],
    'Longitude': [102.2721, 102.2735, 102.2710, 102.2750, 102.2800]
}

df_spasial = pd.DataFrame(data_lokasi)

# Mengubah DataFrame biasa menjadi GeoDataFrame berbasis WGS84 (EPSG:4326)
geometry = [Point(xy) for xy in zip(df_spasial['Longitude'], df_spasial['Latitude'])]
gdf = gpd.GeoDataFrame(df_spasial, geometry=geometry, crs="EPSG:4326")

print("📌 GeoDataFrame Spasial (5 Lokasi Titik):")
print(gdf[['Nama_Fasilitas', 'Kategori', 'geometry']], "\n")

# 4. SPATIAL ANALYSIS & FILTERING
# Spatial Query: Filter lokasi dengan kapasitas > 400 orang
gdf_filtered = gdf[gdf['Kapasitas_Orang'] > 400]

print(f"✅ Filter Spasial: Ditemukan {len(gdf_filtered)} lokasi berkapasitas besar.\n")

# 5. INTEGRASI WEB MAPPING INTERAKTIF (FOLIUM ENGINE)
# Titik Pusat Peta (Universitas Bengkulu)
center_lat, center_lon = -3.7575, 102.2730
m = folium.Map(location=[center_lat, center_lon], zoom_start=15, tiles='OpenStreetMap')

# Menambahkan Layer Marker untuk Setiap Fasilitas
for idx, row in gdf.iterrows():
    # Pemilihan warna marker berdasarkan kategori
    color_map = {
        'Pendidikan': 'blue',
        'Akademik': 'red',
        'Fasilitas': 'green',
        'Olahraga': 'orange',
        'Publik': 'purple'
    }

    popup_text = f"""
    <b>{row['Nama_Fasilitas']}</b><br>
    Kategori: {row['Kategori']}<br>
    Kapasitas: {row['Kapasitas_Orang']} Orang
    """

    folium.Marker(
        location=[row['Latitude'], row['Longitude']],
        popup=folium.Popup(popup_text, max_width=250),
        tooltip=row['Nama_Fasilitas'],
        icon=folium.Icon(color=color_map.get(row['Kategori'], 'gray'), icon='info-sign')
    ).add_to(m)

# Menambahkan Poligon Sederhana (Simulasi Area Kampus Utama)
kampus_polygon = Polygon([
    (102.2690, -3.7550),
    (102.2760, -3.7550),
    (102.2760, -3.7610),
    (102.2690, -3.7610)
])

gdf_boundary = gpd.GeoDataFrame(geometry=[kampus_polygon], crs="EPSG:4326")
folium.GeoJson(
    gdf_boundary,
    style_function=lambda x: {
        'fillColor': '#3186cc',
        'color': '#1a5276',
        'weight': 2,
        'fillOpacity': 0.15
    },
    tooltip="Batas Kawasan Kampus UNIB"
).add_to(m)

# Display Peta Interaktif
print("🗺️ Menampilkan Peta Web Geospasial Interaktif:")
m
```
