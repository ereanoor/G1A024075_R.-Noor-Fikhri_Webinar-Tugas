```python
# ==============================================================================
# INFORMATICS PRACTITIONER LECTURE SERIES 2026 - UNIB
# IMPLEMENTASI MATERI 1: High Performance Computing (HPC) & Parallel Processing
# Pemateri: Asst. Prof. Dr. Worawan Diaz Carballo (Thammasat University, Thailand)
# ==============================================================================

import time
import numpy as np
import multiprocessing as mp

# 1. MENAMPILKAN SPESIFIKASI INFRASTRUKTUR / SPESIFIKASI SIMULASI HPC
def check_hpc_environment():
    num_cpus = mp.cpu_count()
    print("=" * 65)
    print(" 🖥️ SPESIFIKASI INFRASTRUKTUR COMPUTING (GOOGLE COLAB / HPC NODE)")
    print("=" * 65)
    print(f" Jumlah CPU Cores Terdeteksi : {num_cpus} Cores")
    print(" Tipe Pengolahan            : Multi-core Parallel Processing")
    print("=" * 65 + "\n")
    return num_cpus

# 2. FUNGSI SIMULASI BEBAN KOMPUTASI SPASIAL / MATRIKS (FLOATING POINT OPERATIONS - FLOPS)
def compute_chunk(chunk):
    """
    Melakukan operasi matematika numerik bertingkat pada chunk data matriks
    (Mengembangkan perhitungan simulasi data spasial/numerik lokal)
    """
    # Mengalikan matriks dan menghitung akar kuadrat untuk mensimulasikan beban komputasi
    result = np.sin(chunk) ** 2 + np.cos(chunk) ** 2
    return np.sum(result)

# 3. IMPLEMENTASI EKSEKUSI SERIAL (SINGLE-CORE / SEKUANSIAL)
def run_serial_processing(data):
    print("⏳ Menjalankan Komputasi secara SERIAL (1 CPU Core)...")
    start_time = time.time()

    # Eksekusi sekuansial
    total_sum = sum(compute_chunk(chunk) for chunk in data)

    elapsed_time = time.time() - start_time
    print(f"✅ Selesai Serial! Waktu Eksekusi: {elapsed_time:.4f} detik\n")
    return elapsed_time

# 4. IMPLEMENTASI EKSEKUSI PARALEL (MULTI-CORE HPC SIMULATION)
def run_parallel_processing(data, num_cores):
    print(f"⚡ Menjalankan Komputasi secara PARALEL ({num_cores} CPU Cores)...")
    start_time = time.time()

    # Membagi beban kerja ke seluruh core CPU yang tersedia menggunakan Multiprocessing Pool
    with mp.Pool(processes=num_cores) as pool:
        results = pool.map(compute_chunk, data)
        total_sum = sum(results)

    elapsed_time = time.time() - start_time
    print(f"✅ Selesai Paralel! Waktu Eksekusi: {elapsed_time:.4f} detik\n")
    return elapsed_time

# 5. MAIN PROGRAM / EKSEKUSI UTAMA
if __name__ == "__main__":
    # A. Cek Infrastruktur
    cores = check_hpc_environment()

    # B. Penyiapan Dataset Matriks Besar (Simulasi Dataset Beban Lokal)
    print("📦 Membuat Dataset Simulasi Matriks Ukuran Besar...")
    N_CHUNKS = 16
    CHUNK_SIZE = 2_000_000

    # Membagi data menjadi 16 potongan (chunks) berisi angka floating point
    dataset = [np.random.rand(CHUNK_SIZE) for _ in range(N_CHUNKS)]
    print(f" Total Data Points: {N_CHUNKS * CHUNK_SIZE:,} Floating-Point Numbers\n")

    # C. Jalankan Simulasi Serial vs Paralel
    time_serial = run_serial_processing(dataset)
    time_parallel = run_parallel_processing(dataset, num_cores=cores)

    # D. Evaluasi Performa (Speedup Factor & Efisiensi)
    speedup = time_serial / time_parallel
    efficiency = (speedup / cores) * 100

    print("=" * 65)
    print(" 📊 HASIL ANALISIS PERFORMA HIGH PERFORMANCE COMPUTING (HPC)")
    print("=" * 65)
    print(f" Waktu Komputasi Serial  : {time_serial:.4f} detik")
    print(f" Waktu Komputasi Paralel : {time_parallel:.4f} detik")
    print(f" Faktor Percepatan       : {speedup:.2f}x Lebih Cepat")
    print(f" Efisiensi Penggunaan CPU: {efficiency:.2f}%")
    print("=" * 65)
    print("💡 KESIMPULAN:")
    print(" Pemanfaatan komputasi paralel pada infrastruktur HPC terbukti")
    print(" memangkas waktu pemrosesan data numerik berukuran besar secara signifikan.")
```
