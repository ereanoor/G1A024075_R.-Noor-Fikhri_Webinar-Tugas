```python
# ==============================================================================
# INFORMATICS PRACTITIONER LECTURE SERIES 2026 - UNIB
# IMPLEMENTASI MATERI 4: Introduction to Machine Learning (Pipeline & Algorithms)
# Pemateri: Assoc. Prof. Ts. Dr. Taniza Tajuddin (UiTM Kedah, Malaysia)
# ==============================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LinearRegression
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import accuracy_score, classification_report, mean_squared_error, r2_score

# 1. MENAMPILKAN SPESIFIKASI WORKFLOW MACHINE LEARNING
print("=" * 65)
print(" 🤖 WORKFLOW & TAKSONOMI MACHINE LEARNING (DR. TANIZA TAJUDDIN)")
print("=" * 65)
print(" 1. Data Collection & Preprocessing (Cleaning & Feature Scaling)")
print(" 2. Supervised Learning: Classification (e.g., Random Forest)")
print(" 3. Supervised Learning: Regression (e.g., House Price Prediction)")
print(" 4. Unsupervised Learning: Clustering (K-Means) & PCA Reduction")
print(" 5. Model Evaluation Metrics (Accuracy, R2 Score, Confusion Matrix)")
print("=" * 65 + "\n")

# ------------------------------------------------------------------------------
# A. SUPERVISED LEARNING: KLASIFIKASI (DETEKSI SPAM / KATEGORI)
# ------------------------------------------------------------------------------
print("📌 [BAGIAN 1] SUPERVISED LEARNING - KLASIFIKASI DATA")
np.random.seed(42)
n_samples = 300

# Generating dataset (Fitur: Durasi Belajar, Kehadiran, Nilai Tugas -> Target: Lulus/Gagal)
study_hours = np.random.uniform(1, 10, n_samples)
attendance = np.random.uniform(50, 100, n_samples)
assignment_scores = np.random.uniform(40, 100, n_samples)

# Target label logika bisnis
passed = np.where((study_hours > 4) & (attendance > 70) & (assignment_scores > 60), 1, 0)

df_class = pd.DataFrame({
    'Jam_Belajar': study_hours,
    'Persen_Kehadiran': attendance,
    'Nilai_Tugas': assignment_scores,
    'Status_Lulus': passed
})

X_c = df_class[['Jam_Belajar', 'Persen_Kehadiran', 'Nilai_Tugas']]
y_c = df_class['Status_Lulus']

X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(X_c, y_c, test_size=0.2, random_state=42)

# Training Classifier (Random Forest)
clf = RandomForestClassifier(n_estimators=50, random_state=42)
clf.fit(X_train_c, y_train_c)
y_pred_c = clf.predict(X_test_c)

print(f" ✅ Akurasi Model Klasifikasi: {accuracy_score(y_test_c, y_pred_c) * 100:.2f}%")
print(" Classification Report:")
print(classification_report(y_test_c, y_pred_c, target_names=['Gagal', 'Lulus']))

# ------------------------------------------------------------------------------
# B. SUPERVISED LEARNING: REGRESI (PREDIKSI HARGA RUMAH)
# ------------------------------------------------------------------------------
print("\n📌 [BAGIAN 2] SUPERVISED LEARNING - REGRESI (PREDIKSI HARGA)")

# Fitur: Luas Rumah (sqft), Jumlah Kamar -> Target: Harga (Rupiah/USD)
house_size = np.random.uniform(500, 3500, n_samples)
num_bedrooms = np.random.randint(1, 6, n_samples)
price = (house_size * 150) + (num_bedrooms * 10000) + np.random.normal(0, 15000, n_samples)

df_reg = pd.DataFrame({'Luas_Rumah': house_size, 'Jumlah_Kamar': num_bedrooms, 'Harga': price})

X_r = df_reg[['Luas_Rumah', 'Jumlah_Kamar']]
y_r = df_reg['Harga']

X_train_r, X_test_r, y_train_r, y_test_r = train_test_split(X_r, y_r, test_size=0.2, random_state=42)

reg = LinearRegression()
reg.fit(X_train_r, y_train_r)
y_pred_r = reg.predict(X_test_r)

print(f" ✅ Skor R2 (Akurasi Regresi): {r2_score(y_test_r, y_pred_r):.4f}")
print(f" Mean Squared Error (MSE): {mean_squared_error(y_test_r, y_pred_r):,.2f}")

# ------------------------------------------------------------------------------
# C. UNSUPERVISED LEARNING: CLUSTERING (K-MEANS)
# ------------------------------------------------------------------------------
print("\n📌 [BAGIAN 3] UNSUPERVISED LEARNING - K-MEANS CLUSTERING")

scaler = StandardScaler()
X_scaled_c = scaler.fit_transform(X_c)

# Mengelompokkan data siswa menjadi 3 Klaster tanpa Label
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
df_class['Cluster'] = kmeans.fit_predict(X_scaled_c)

print(" Distribusi Anggota Tiap Klaster:")
print(df_class['Cluster'].value_counts())

# Visualisasi PCA Reduksi Dimensi + Clustering Plot
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled_c)

plt.figure(figsize=(7, 5))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=df_class['Cluster'], cmap='viridis', alpha=0.7, edgecolors='k')
plt.title('Unsupervised Learning: K-Means Clustering dengan PCA (2D)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.colorbar(label='Cluster ID')
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

print("\n💡 KESIMPULAN MATERI 4:")
print(" Pembelajaran Machine Learning mencakup penguasaan Supervised Learning (Klasifikasi/Regresi)")
print(" dan Unsupervised Learning (Clustering/PCA) untuk ekstraksi wawasan cerdas berbasis data.")
```
